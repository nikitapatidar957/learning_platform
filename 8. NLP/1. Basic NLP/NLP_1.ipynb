## Natural Language Processing With Python's NLTK Package

### Tokenizing

```python
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
example_string = """
Muad'Dib learned rapidly because his first training was in how to learn.
And the first lesson of all was the basic trust that he could learn.
It's shocking to find how many people do not believe they can learn,
and how many more believe learning to be difficult."""
```

```python
sent_tokenize(example_string)
```

```python
word_tokenize(example_string)
```

### Filtering Stop Words

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
## Stopword stored in list
stop_words = set(stopwords.words("english"))
```

```python
##Example
worf_quote = "Sir, I protest. I am not a merry man!"

## Tokenized the sentence by words
words_in_quote = word_tokenize(worf_quote)
words_in_quote
```

```python
###  create an empty list to hold the words that make it past the filter:
filtered_list = []

### filtered_list, to hold all the words in words_in_quote that aren’t stop words.
## Now you use stop_words to filter words_in_quote:
for word in words_in_quote:
    if word.casefold() not in stop_words:
        filtered_list.append(word)
```

```python
filtered_list
```

### Stemming

```python
## Import the relevant parts of NLTK in order to start stemming:
from nltk.stem import PorterStemmer

## create a stemmer with PorterStemmer()
stemmer = PorterStemmer()
```

```python
## Example for stemming

string_for_stemming = """The crew of the USS Discovery discovered many discoveries. Discovering is what explorers do."""
```

```python
words = word_tokenize(string_for_stemming)
words
```

```python
stemmed_words = [stemmer.stem(word) for word in words]
stemmed_words
```

### Lemmatizing

```python
## import the relevant parts of NLTK in order to start lemmatizing
from nltk.stem import WordNetLemmatizer

## Create a lemmatizer to use:
lemmatizer = WordNetLemmatizer()
```

```python
## lemmatizing a plural noun
lemmatizer.lemmatize("scarves")
```

```python
## Sentence for Lemmatization
string_for_lemmatizing  = """The crew of the USS Discovery discovered many discoveries. Discovering is what explorers do."""
```

```python
## Tokenizing Words
words = word_tokenize(string_for_lemmatizing)
words
```

```python
lemmatized_words = [lemmatizer.lemmatize(word) for word in words]
lemmatized_words
```

```python
## lemmatized a word that looked very different from its lemma, Eg "Worst"
lemmatizer.lemmatize("worst")
```

The result 'worst' because lemmatizer.lemmatize() assumed that "worst" was a noun. We can make it clear that you want "worst" to be an adjective:

```python
lemmatizer.lemmatize("worst", pos="a")
```
