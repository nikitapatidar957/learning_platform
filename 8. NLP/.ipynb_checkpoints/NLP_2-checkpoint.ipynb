## Create a Bag of Words Model with Sklearn

```python
## importing basic libraries
import pandas as pd
import numpy as np
import collections
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
 
doc1 = 'Game of Thrones is an amazing tv series!'
doc2 = 'Game of Thrones is the best tv series!'
doc3 = 'Game of Thrones is so great'
 
##  instantiate a CountVectorizer object  
CountVec = CountVectorizer()    

#transform
Count_data = CountVec.fit_transform([doc1,doc2,doc3])
 
#create dataframe
cv_dataframe=pd.DataFrame(Count_data.toarray(),columns=CountVec.get_feature_names())
print(cv_dataframe)
```

## Feature Extraction with Tf-Idf vectorizer

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
 
doc1 = 'Game of Thrones is an amazing tv series!'
doc2 = 'Game of Thrones is the best tv series!'
doc3 = 'Game of Thrones is so great'
 
vectorizer = TfidfVectorizer()
vectors = vectorizer.fit_transform([doc1,doc2,doc3])
feature_names = vectorizer.get_feature_names()

## As the output in sparse matrix convert to df
dense = vectors.todense()
denselist = dense.tolist()
df = pd.DataFrame(denselist, columns=feature_names)
df
```

## Word Embedding through Word2Vec with Gensim Library

```python
import bs4 as bs
import urllib.request
import re
import nltk

## Creating Corpus
scrapped_data = urllib.request.urlopen('https://en.wikipedia.org/wiki/Artificial_intelligence')
article = scrapped_data .read()

## read the article content and parse it using an object of the BeautifulSoup class
parsed_article = bs.BeautifulSoup(article,'lxml')

##  find_all function of the BeautifulSoup object to fetch all the contents from the paragraph tags of the article.
paragraphs = parsed_article.find_all('p')

article_text = ""

for p in paragraphs:
    article_text += p.text
    
print(article_text)
```

```python
# Cleaning the text
## we convert all the text to lowercase and then remove all the digits, special characters, and extra spaces from the text.

processed_article = article_text.lower()
processed_article = re.sub('[^a-zA-Z]', ' ', processed_article )
processed_article = re.sub(r'\s+', ' ', processed_article)

# Preparing the dataset
all_sentences = nltk.sent_tokenize(processed_article)

all_words = [nltk.word_tokenize(sent) for sent in all_sentences]

# Removing Stop Words
from nltk.corpus import stopwords
for i in range(len(all_words)):
    all_words[i] = [w for w in all_words[i] if w not in stopwords.words('english')]
```

```python
## Creating Word2Vec Model

from gensim.models import Word2Vec

word2vec = Word2Vec(all_words, min_count=2)        
## value of 2 for min_count specifies to include only those words in the Word2Vec model that appear at least twice in the corpus
```

```python
## To see the dictionary of unique words that exist at least twice in the corpus, execute the following script:

vocabulary = word2vec.wv.index_to_key
vocabulary
```

### Model Analysis

```python
## Finding Vectors for a Word
v1 = word2vec.wv['artificial']
v1

### By default, a hundred dimensional vector is created by Gensim Word2Vec
# If we use the bag of words approach for embedding the article, the length of the vector for each will be 1206 since 
# there are 1206 unique words with a minimum frequency of 2
```

```python
## Finding Similar Words
sim_words = word2vec.wv.most_similar('intelligence')
sim_words 
```
